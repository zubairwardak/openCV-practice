```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("filter.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap='gray')
```

```python
x = -10

horizontal_kernel = np.array([[-1,-1,-1],
                              [x, x,  x],
                              [1, 1,  1]])

vertical_kernel = np.array([[-1, x, 1],
                            [-1, x, 1],
                            [-1, x, 1]])

digonal_kernel = np.array([[x,-1,-1],
                           [1, x,-1],
                           [1, 1,x]])
horizontal_filter_image = cv2.filter2D(image, -1, horizontal_kernel)
vertical_filter_image = cv2.filter2D(image, -1, vertical_kernel)
digonal_filter_image = cv2.filter2D(image , -1, digonal_kernel)

plt.Figure(figsize=(12,12))

plt.subplot(2,2,1)
plt.title("orignal Image")
plt.imshow(image, cmap='gray')

plt.subplot(2,2,2)
plt.title("Horizontal image")
plt.imshow(horizontal_filter_image, cmap='gray')

plt.subplot(2,2,3)
plt.title("Vertical image")
plt.imshow(vertical_filter_image, cmap="gray")

plt.subplot(2,2,4)
plt.title("Digonal image")
plt.imshow(digonal_filter_image)


plt.show()
```
