```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./giraffe.webp")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
print(image.shape)
print(image_rgb.shape)
plt.imshow(image_rgb)
```

```python
# Resizing the image to a specific size
resized_image = cv2.resize(image_rgb, (400, 400))
```

```python
print(resized_image.shape)
plt.imshow(resized_image)
plt.title("Resized Image")
```

# Scale image 

```python
scaled_percentage = 50
s_width = int(image_rgb.shape[1] * scaled_percentage / 100)
s_height = int(image_rgb.shape[0] * scaled_percentage / 100)

scaled_image = cv2.resize(image_rgb, (s_width, s_height) , interpolation = cv2.INTER_AREA)
```

```python
print(scaled_image.shape)
plt.imshow(resized_image)
plt.title("Scale Image")
```

# Interpolation 

```python
resize_width = 600
resize_height = 600

resize_rgb_image = cv2.resize(image_rgb, (200, 200))

methods = {
    "NEAREST": cv2.INTER_NEAREST,
    "LINEAR": cv2.INTER_LINEAR,
    "CUBIC": cv2.INTER_CUBIC,
    "AREA": cv2.INTER_AREA,
    "LANCZOS4": cv2.INTER_LANCZOS4

}

resized_images = {}
for name, method in methods.items():
    resized_images[name] = cv2.resize(resize_rgb_image, (resize_width, resize_height), interpolation=method)
    cv2.imwrite(f"./{name}.jpg", resized_images[name])

plt.figure(figsize=(10, 8))

#orginal image
plt.subplot(2, 3, 1)
plt.imshow(resize_rgb_image)
plt.title("Original Image")
plt.axis("off")

# Resized images with different interpolation methods
for i, (name, resized_img) in enumerate(resized_images.items(), start=2):
    plt.subplot(2, 3, i)
    plt.imshow(resized_img)
    plt.title(name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
