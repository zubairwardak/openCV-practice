```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./giraffe.webp")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb)

```

# Flip

```python
flip_horizontal = cv2.flip(image_rgb, 1)
filp_vertical = cv2.flip(image_rgb, 0)
flip_both = cv2.flip(image_rgb, -1)

title = ["Original Image", "Flip Horizontal", "Flip Vertical", "Flip Both"]
images = [image_rgb, 
            flip_horizontal,
            filp_vertical,
            flip_both
          ]

plt.figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i])
    plt.title(title[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Rotate

```python
rotate_90_clockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_CLOCKWISE)
rotate_90_counterclockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)
rotate_180 = cv2.rotate(image_rgb, cv2.ROTATE_180)


titles = [ "Orignal Image", "Rotate_90_Clockwise", "Rotate_90_CounterClockwise", "Rotate_180" ]

images = [image_rgb,
            rotate_90_clockwise,
            rotate_90_counterclockwise,
            rotate_180
          ]

plt.figure(figsize=(12, 8))
for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Crop the image 

```python
x_start, y_start = 400, 300
x_end, y_end = 800, 700

cropped_image = image_rgb[y_start:y_end, x_start:x_end]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cropped_image)
plt.title("Cropped Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
