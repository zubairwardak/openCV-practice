```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./butterfly.jpg")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
```

```python
plt.imshow(image_rgb)
```

```python
lower_bound = np.array([0, 0,190])
upper_bound = np.array([150, 200, 255])
```

```python
mask = cv2.inRange(image_hsv, lower_bound, upper_bound)
```

```python
plt.imshow(mask)
```

```python
black_hsv = np.uint8([[[120, 255, 0]]])
black_bgr = cv2.cvtColor(black_hsv, cv2.COLOR_HSV2BGR)
```

```python
image_result = image.copy()
image_result[mask > 0] = black_bgr
```

```python
image_result_rgb = cv2.cvtColor(image_result, cv2.COLOR_BGR2RGB)
```

```python
fig, axi = plt.subplots(1, 3, figsize=(15, 5))

axi[0].imshow(image_rgb)
axi[0].set_title("Original Image")
axi[1].axis("off")

axi[1].imshow(mask)
axi[1].set_title("Binary Mask")
axi[1].axis("off")

axi[2].imshow(image_result_rgb)
axi[2].set_title("Result Image")
axi[2].axis("off")

plt.tight_layout()
plt.show()
```
