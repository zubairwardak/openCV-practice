# Creating Zeros Tensors

```python
import torch
```

```python
tensor_zero = torch.zeros(5)
print(tensor_zero)
```

```python
tensor_zero_2d = torch.zeros(256, 256)
print(tensor_zero_2d)
print(tensor_zero_2d.shape)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(tensor_zero_2d.numpy(), cmap='grey', vmin=0, vmax=1)
plt.title("Black Image")
plt.axis("off")
plt.show()
```

# Creating Ones Tensors

```python
tensor_one = torch.ones(5)
print(tensor_one)
```

```python
tensor_one = torch.ones(256, 256)
print(tensor_one)
print(tensor_one.shape)
print(tensor_one.ndim)
```

```python
plt.imshow(tensor_one.numpy(), cmap='grey', vmin=0, vmax=1)
plt.title("White image")
plt.axis("off")
plt.show()
```
