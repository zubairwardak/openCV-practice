```python
import torch
```

```python
tensor1 =torch.tensor([1, 2, 3])
tensor2 =torch.tensor([4, 5, 6])
```

```python
stacked_tensor_dim0 = torch.stack((tensor1, tensor2), dim=0)
print("Shape : ", stacked_tensor_dim0.shape)
print(stacked_tensor_dim0)

print("+"*30)

stacked_tensor_dim1 = torch.stack((tensor1, tensor2) , dim=1)
print("Shape : ", stacked_tensor_dim1.shape)
print(stacked_tensor_dim1)
```

```python
tensor3 = torch.tensor([7, 8, 9])

stacked_tensor_dim0_new = torch.stack((tensor1, tensor2, tensor3), dim= 0)
print("Shape : ", stacked_tensor_dim0_new.shape)
print(stacked_tensor_dim0_new)
```

```python
tensor_1 = torch.tensor([[1,2],[3,4]])
tensor_2 = torch.tensor([[5,6],[7,8]])
print("dimension :", tensor_1.ndim)
stacked_tensor_2d  = torch.stack((tensor_1, tensor_2), dim=0)
print("Shape : ", stacked_tensor_2d.shape)
print("dimension : ", stacked_tensor_2d.ndim)
print(stacked_tensor_2d)
```

```python
stacked_tensor_2d_dim1  = torch.stack((tensor_1, tensor_2), dim=1)
print("Shape : ", stacked_tensor_2d_dim1.shape)
print("dimension : ", stacked_tensor_2d_dim1.ndim)
print(stacked_tensor_2d_dim1)
```

```python
stacked_tensor_2d_dim2  = torch.stack((tensor_1, tensor_2), dim=2)
print("Shape : ", stacked_tensor_2d_dim2.shape)
print("dimension : ", stacked_tensor_2d_dim2.ndim)
print(stacked_tensor_2d_dim2)
```

```python
concatenated_tensor_dim0 = torch.cat((tensor_1, tensor_2), dim=0)
print("Shape : ", concatenated_tensor_dim0.shape)
print("Dimension : ", concatenated_tensor_dim0.ndim)
print(concatenated_tensor_dim0)
```

```python
concatenated_tensor_dim1 = torch.cat((tensor_1, tensor_2), dim=1)
print("Shape : ", concatenated_tensor_dim1.shape)
print("Dimension : ", concatenated_tensor_dim1.ndim)
print(concatenated_tensor_dim1)
```
