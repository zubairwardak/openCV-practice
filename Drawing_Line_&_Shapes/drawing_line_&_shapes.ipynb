```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

# Drawing Canvas

```python
width, height = 700, 500
blue = (255, 0 , 0)
canvas = np.full((height, width, 3), blue, dtype = np.uint8)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Drawing Line

```python
cv2.line(canvas, (50,50), (600, 450), (0,255,0), 5)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Drawing Rectangle

```python
cv2.rectangle(canvas, (70,70), (350, 300), (0,255,255), 1)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Drawing Circle 

```python
cv2.circle(canvas, (500, 200), 100, (0,0,255), 2)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
cv2.ellipse(canvas, (200, 400), (100, 50), 0, 0, 360, (255,255,0), 2)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

 # Drawing Polygone

```python
points = np.array([[150, 400], [300, 400], [225, 300]], np.int32)
new_points = points.reshape((-1, 1, 2))
cv2.polylines(canvas, [new_points], True, (255,0,255), 2)
plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```
