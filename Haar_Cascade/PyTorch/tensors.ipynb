```python
import torch
```

```python
# Scalar tensor 0D
scalar = torch.tensor(42)
print("Scalar Tensor : ", scalar)
print("Tensor Dimintional : ", scalar.ndim)
print("Shape of Scalar : ", scalar.shape)
```

```python
# vetor : 1D tensor

vector = torch.tensor([1,2,3])
print("Vector Tensor : ", vector)
print("Vector Daiminsion : ", vector.ndim)
print("Shape of Vector : ", vector.shape)
```

```python
# Matrix 2D 
matrix = torch.tensor([[1,2],[4,5]])
print("Matrix Tensor : ", matrix)
print("Matrix Daiminsion : ", matrix.ndim)
print("Shape of Matrix : ", matrix.shape)
```
