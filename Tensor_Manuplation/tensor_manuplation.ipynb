```python
import torch
```

```python
orignal_tensor = torch.arange(12)
print(orignal_tensor)
print(orignal_tensor.shape)
print(orignal_tensor.nelement())
print(orignal_tensor.ndim)
```

```python
reshaped_tensor = orignal_tensor.reshape(3,4)
print(reshaped_tensor)
print(reshaped_tensor.shape)
print(reshaped_tensor.nelement())
print(reshaped_tensor.ndim)
```

# Tensor merging 


```python
tensor1 = torch.tensor([[1, 2], [3, 4]])
tensor2 = torch.tensor([[5, 6], [7, 8]])

concat_tensor_row = torch.cat((tensor1, tensor2), dim=0)
concat_tensor_col = torch.cat((tensor1, tensor2), dim=1)
print(concat_tensor_row)
print(concat_tensor_col)
```

# Splitting 

```python
original_tensor = torch.arange(12)
chunks = torch.chunk(orignal_tensor, 6, dim=0)
print(chunks)

for chunk in chunks:
    print(chunk)
```

```python
# splitting 

splitted_tensors = torch.split(orignal_tensor, 5, dim=0)
print(splitted_tensors)

for split in splitted_tensors:
    print(split)
```
