```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("giraffe.webp")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_grey = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
kernel_s = 5
average_blur = cv2.blur(image_grey, (kernel_s, kernel_s))
plt.imshow(average_blur, cmap='grey')
```

```python
kernel_size = 31 
gaussian_blur = cv2.GaussianBlur(image_grey, (kernel_size, kernel_size), 0)
plt.imshow(gaussian_blur , cmap='grey')
```

```python
median_blur = cv2.medianBlur(image_grey, 31)
plt.imshow(median_blur , cmap='grey')
```

```python
plt.figure(figsize=(10, 10 ))

plt.subplot(2,2,1)
plt.imshow(image , cmap="grey")
plt.title("Orignal Image")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(average_blur , cmap="grey")
plt.title("Average Blur")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(gaussian_blur , cmap="grey")
plt.title("Gaussian Blur")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(median_blur , cmap="grey")
plt.title("Median Blur")
plt.axis('off')

plt.show()
```
